```python
import sys
import os
from pathlib import Path

# Find the absolute path to your project root (Draci_zvesti_merged)
# This assumes the notebook is in neat_ai/analysis/notebooks/
project_root = Path.cwd().resolve()
while project_root.name != 'Draci_zvesti_merged' and project_root.parent != project_root:
    project_root = project_root.parent

if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))
    
print(f"Project root added to path: {project_root}")
```

```python
import plotly.express as px
from neat_ai.analysis.card_metrics import get_card_power_rating

# Load the power ratings
power_df = get_card_power_rating()

if not power_df.empty:
    # Create a Scatter Plot comparing Power, Win Rate, and Popularity
    fig = px.scatter(
        power_df, 
        x='times_played', 
        y='average_score', 
        size='power_rating',
        color='power_rating',
        hover_name='card_id',
        title="General Card Power Rating Matrix",
        labels={
            'times_played': 'Popularity (Times Picked by AI)', 
            'average_score': 'Win Rate / Fitness Contribution',
            'power_rating': 'Overall Power Rating (0-100)'
        },
        color_continuous_scale="Viridis"
    )

    # Add quadrant lines to help read the meta
    fig.add_hline(y=power_df['average_score'].median(), line_dash="dot", line_color="red", opacity=0.5)
    fig.add_vline(x=power_df['times_played'].median(), line_dash="dot", line_color="red", opacity=0.5)

    fig.show()
else:
    print("Not enough data to calculate power ratings yet. Run more training generations!")
```

```python
import plotly.express as px
from neat_ai.analysis.card_metrics import get_card_tier_list

# Load the data
tier_list = get_card_tier_list()

if not tier_list.empty:
    # Filter out cards that were barely played to avoid skewed data (flukes)
    tier_list = tier_list[tier_list['times_played'] > 5] 

    # Create a bar chart
    fig = px.bar(
        tier_list, 
        x='card_id', 
        y='average_score', 
        color='average_score',
        title="Card Tier List by AI Fitness (Buff / Nerf Targets)",
        labels={'average_score': 'Average Fitness', 'card_id': 'Card Name'},
        color_continuous_scale="Turbo"
    )
    fig.show()
else:
    print("No card data available yet.")
```

```python
import plotly.graph_objects as go
import numpy as np
from neat_ai.analysis.matchups import get_matchup_matrix_with_cards

matrix, card_map = get_matchup_matrix_with_cards()

if not matrix.empty:
    # Build text matrices matching the grid layout to feed into hover metrics
    deck1_list = matrix.index.tolist()
    deck2_list = matrix.columns.tolist()
    
    hover_text = []
    for d1 in deck1_list:
        row_hovers = []
        for d2 in deck2_list:
            d1_cards = card_map.get(d1, "Unknown Card List")
            d2_cards = card_map.get(d2, "Unknown Card List")
            # Format text cleanly with HTML line-breaks
            row_hovers.append(
                f"<b>Playing As Deck ID:</b> {d1}<br>"
                f"<b>Playing As Composition:</b> {d1_cards}<br><br>"
                f"<b>Opponent Deck ID:</b> {d2}<br>"
                f"<b>Opponent Composition:</b> {d2_cards}"
            )
        hover_text.append(row_hovers)

    fig = go.Figure(data=go.Heatmap(
        z=matrix.values,
        x=deck2_list,
        y=deck1_list,
        text=hover_text,
        hoverinfo='text+z', # Tells Plotly to render our clean HTML text block alongside raw score
        colorscale='RdYlGn',
        colorbar=dict(title="Avg G1 Score")
    ))

    fig.update_layout(
        title="Interactive Meta Matchup Heatmap (Hover to see Full Deck Card Compositions)",
        xaxis_title="Opponent Deck ID",
        yaxis_title="Playing As Deck ID"
    )
    fig.show()
else:
    print("No matchup data available yet.")
```

```python
import plotly.express as px
from neat_ai.analysis.learning_curves import get_high_skill_decks
from neat_ai.analysis.data_loader import load_master_df

high_skill = get_high_skill_decks()

if not high_skill.empty:
    print("Top High-Skill / High-Potential Archetypes:")
    # Drop the long hash signature column for the markdown terminal printout 
    # so humans can easily scan card lists in the table
    display(high_skill[['composition', 'learning_slope', 'fitness_delta', 'gens_active']].head(5))

    # Pull the highest slope archetype
    best_match = high_skill.iloc[0]
    best_sig = best_match['signature']
    cards_list = best_match['composition']
    
    df_master = load_master_df()
    deck_trajectory = df_master[df_master['signature'] == best_sig].groupby('gen')['score'].mean().reset_index()
    deck_trajectory['deck_composition'] = cards_list # Bind the list to dataframe for hover extraction

    fig = px.line(
        deck_trajectory, 
        x='gen', 
        y='score', 
        hover_data=['deck_composition'],
        title=f"Learning Curve Trajectory<br><sup>Archetype: {cards_list[:90]}...</sup>",
        labels={'gen': 'Generation Number', 'score': 'Average Evaluated Fitness', 'deck_composition': 'Composition'},
        markers=True
    )
    fig.show()
else:
    print("Not enough generational data to evaluate curves yet.")
```

```python
import plotly.express as px
from neat_ai.analysis.synergies import get_card_synergies

synergies = get_card_synergies(percentile=0.85)

if not synergies.empty:
    top_combos = synergies.head(25).copy()
    
    # Create an explicit human-readable pairing descriptor label for the axis
    top_combos['Card Combo Pairing'] = top_combos['card_A'] + " + " + top_combos['card_B']

    fig = px.bar(
        top_combos,
        x='co_occurrences',
        y='Card Combo Pairing',
        color='co_occurrences',
        orientation='h',
        title="Top 25 Most Frequent Card Synergy Pairings Discovered inside Elite Decks (Top 15%)",
        labels={'co_occurrences': 'Total Concurrent Deployments across Successful Generations', 'Card Combo Pairing': 'Synergy Pair'},
        color_continuous_scale="Cividis"
    )
    
    # Force sorting layout hierarchy
    fig.update_layout(yaxis={'categoryorder':'total ascending'}, coloraxis_showscale=False)
    fig.show()
else:
    print("No synergy metrics compiled yet. Ensure elite threshold parameters match active database score counts.")
```
